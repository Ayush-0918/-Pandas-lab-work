```python
'''
5.create a dataframe with column name,marks1,marks2,marks3 perfrom the following task

*add new column total containg total marks
*add new column average containg the average marks
*add a column result where student with avg>=40 are "pass" orelse "fail"
*display stuents who have passed
*print teh final dataframe

@10:17
'''
```

```python
data={'name':['amit','riya','karan','neha','vikas'],
      'marks1':[70,45,30,80,55],
      'marks2':[65,50,35,75,60],
      'marks3':[75,40,25,85,50]}
df=pd.DataFrame(data)
print(df)
```

```python
df['total']=df['marks1']+df['marks2']+df['marks3']
```

```python
df['avg']=df['total']/3
```

```python
df['result']=df['avg'].apply(lambda x:'pass' if x>=40 else "fail")
```

```python
df[df['result']=='pass']
```
